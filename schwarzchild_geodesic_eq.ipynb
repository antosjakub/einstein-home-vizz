```python
import sympy
from sympy import symbols, Matrix, diag, sin, Function, Eq, diff
```

```python
d = 4
t, r, theta, phi = symbols("t r theta varphi")
coords = [t,r,theta,phi]
M = symbols("M")
```

```python
f = 1 - 2*M/r
```

```python
g = Matrix(diag(-f, 1/f, r**2, r**2 * sin(theta)**2))
g_inv = g.inv()
```

```python
christ_1 = [[[0 for _ in range(d)] for _ in range(d)] for _ in range(d)]
christ_2 = [[[0 for _ in range(d)] for _ in range(d)] for _ in range(d)]
```

```python
# christ 3 lower
for gamma in range(d):
    for alpha in range(d):
        for beta in range(d):
            christ_1[gamma][alpha][beta] = sympy.Rational(1,2) * (
                g[gamma,alpha].diff(coords[beta]) + 
                g[beta,gamma].diff(coords[alpha]) -
                g[alpha,beta].diff(coords[gamma])
            )
# christ 1 upper 2 lower
for sigma in range(d):
    for alpha in range(d):
        for beta in range(d):
            tot = 0
            for gamma in range(d):
                tot += g_inv[sigma,gamma] * christ_1[gamma][alpha][beta]
            christ_2[sigma][alpha][beta] = tot
```

```python
tau = symbols('tau')
t_f = Function('t')(tau)
r_f = Function('r')(tau)
theta_f = Function('theta')(tau)
phi_f = Function('varphi')(tau)
coords_f = [t_f, r_f, theta_f, phi_f]
```

```python
# construct the geodesic equation
geodesic_eqs = []
geodesic_eqs_lhs_terms = []
for gamma in range(d):
    lhs = coords_f[gamma].diff(tau,2)
    geodesic_eqs_lhs_terms.append([])
    for mu in range(d):
        for nu in range(d):
            term = christ_2[gamma][mu][nu]*coords_f[mu].diff()*coords_f[nu].diff()
            lhs += term
            geodesic_eqs_lhs_terms[gamma].append(term)
    geodesic_eqs.append(Eq(lhs, 0))
```

$$
(t)\implies\quad\frac{d}{d\tau}\left[\left(1 - \frac{2M}{r}\right)\frac{dt}{d\tau}\right] = 0 \\
(\varphi)\implies\quad\frac{d}{d\tau} \left[r^2 \sin^2(\theta) \frac{d\varphi}{d\tau}\right] = 0 \\
$$
Also, radially symmetric, so a particle won't deviate from the plane of its orbit, BUNO set
$$
\theta(\tau)=\frac{\pi}{2}
$$

We then label the conserved quantities
$$
E:=\left(1 - \frac{2M}{r}\right)\frac{dt}{d\tau} \\
L:=r^2\frac{d\varphi}{d\tau}
$$

$$
g_{\mu\nu}u^\mu u^\nu = -1
$$

$$
-m^2 = g_{\mu\nu}p^\mu p^\nu = 
g_{tt}(p^t)^2+
g_{rr}(p^r)^2+ 
g_{\theta\theta}(p^\theta)^2+
g_{\varphi\varphi}(p^\varphi)^2
$$

$$
(p^r)^2 = E^2 - \left(1 - \tfrac{2M}{r}\right)\left(m^2 + \tfrac{L^2}{r^2}\right)
$$

As an ODE integration problem:
 - $\tilde E, \tilde L$
 - $r(0)=r_0$
$$
(\tfrac{dr}{d\tau})^2 = \tilde E^2 - \left(1 - \tfrac{2M}{r}\right)\left(1 + \tfrac{\tilde L^2}{r^2}\right) \\
(\tfrac{dr}{d\tau})^2 = \tilde E^2 - \tilde V^2(r)
$$

$$
\tfrac{dr}{d\tau} = \pm\sqrt{\tilde E^2 - \tilde V^2(r)}
$$

$$
\pm \int \frac{dr}{\sqrt{\tilde E^2 - \tilde V^2(r)}} = \int d\tau
$$

$$
\pm \int^r_{r_0} \frac{dr'}{\sqrt{\tilde E^2 - \tilde V^2(r')}} = \int^{\tau}_0 d\tau' = \tau
$$
